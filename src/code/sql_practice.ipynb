```python
import sqlite3
import os
import pandas as pd
```

```python
# データベースに接続
db_path = "../data/sample.db"
conn = sqlite3.connect(db_path)

print(f"データベース作成完了: {db_path}")
```

## SQL操作の基本サイクル

**SQLをPythonから実行するときは、基本的に以下の3ステップで進めます。**

1. Cursor（カーソル）を作る: SQLを実行するための「手」のようなものです。
2. Execute（実行）: SQL文を送り出します。
3. Commit（確定）: 変更を保存します（データ読み取り時は不要）。

```python
# カーソルの作成
cur = conn.cursor()

# 1. テーブルの作成 (usersという名前の表を作る)
cur.execute("""
CREATE TABLE IF NOT EXISTS users (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    age INTEGER
)
""")

# 2. データの挿入
cur.execute("INSERT INTO users (name, age) VALUES ('田中太郎', 25)")
cur.execute("INSERT INTO users (name, age) VALUES ('佐藤花子', 30)")
cur.execute("INSERT INTO users (name, age) VALUES ('鈴木一郎', 22)")

# 3. 変更を保存（重要！）
conn.commit()

print("テーブル作成とデータ挿入が完了しました。")
```

```python
# SQLでデータを取り出して、PandasのDataFrameとして表示
query = "SELECT * FROM users"
df = pd.read_sql(query, conn)

# 表示
df
```
